## 서브워브(SUBWORD LEVEL) 기반 토큰화 - BPE, Wordpiece, unigram
* 트랜스포머 기반 모델(bert, gpt, 라마)에서 표준으로 사용함.
* 단어를 기준으로 토큰x, 문자와 단어의 중간 수준 ➔ 서브워드로 토큰화
* 동작원리 : 
    * 자주 등장하는 문자열 조합 ( 빈도가 낮거나 희귀) ➔ 여러 서브워드로 분할 ➔ 하나의 토큰으로
    * 빈도가 높은단어 ➔ 하나의 토큰
* 장점: 
    * 글자 단위로 붙임 ➔ oov가 나오지 않음. 밥을 ➔ 밥, 을, 밥을 ➔ oov가 나오지 않도록 정의함
    * 어휘 사전 크기 최적화
    * 형태론적 의미 포착 ➔ 접두사 접미사 등. ➔ 문법적으로 이해할 수 있는 방식

### kiwipiepy (korean intelligent word identifier)
- 최근에 많이 사용되는 한국어 토크나이저.
- 체언(명사, 대명사): `N`, 용언(동사, 형용사): `V`<br> 수식언(관형사, 부사): `M`,  관계언(조사):`J`, 어미: `E`, 기호: `S`
---
**Kiwi 클래스 사용** <br>
- 키위 - 자체의 어휘사전을 만들어둠. ( 토큰화할 때 사용할 각각의 단어들을 정의해둠)
- 주요 메소드
    - `tokenize(text)` : 형태소 분석 기반 토큰화 수행 
    - `analyse(text)`: 토크나이즈 보다 좀 더 상세한 분석 진행 & 여러 분석 결과 조회가능 
    - `add_user_word(word, pos, score)`: 사전에 직접 단어 등록
    - `space(text)` : 띄어쓰기 교정


```python
import kiwipiepy
from kiwipiepy import Kiwi
from pprint import pprint

kiwipiepy.__version__
```

```python
kiwi = Kiwi()
```

```python
# 형태소 기반 토큰화 - tokenize()

txt = "나는 자연어 처리를 공부합니다. \n 자연어 처리란 무엇일까요?"
tokens = kiwi.tokenize(txt)

print(tokens) 
```

지금 토큰이라는 객체에 담겨져서 나옴 ➔ token이라는 클래스를 만들어서 묶어서
즉 하나의 값이 여러개의 값으로 구성된 경우 ➔ 자료구조 방법, 클래스를 이용해서 객체로 ➔ 객체로 담겨져서 온다면 값이 여러개존재한다는 거고, 각 객체에 대한 메서드가 존재한다는 것 <br>
Token 객체에 담겨서 리스트로 반환
- form : 단어 정보
- tag : 품사 정보
- start : 시작 인덱스
- len : 글자수 .? 

```python
# Token 객체에서 속성값들 조회

for token in tokens:
    # print(token)
    # token = Token
    result = (f"토큰문자열: {token.form} |"
              f"토큰 원형 : {token.lemma} |"
              f"토큰 품사: {token.tag} |"
              f"문서 내 시작 위치: {token.start} |"
              f"토큰 글자수: {token.len}"
              f"토큰이 있는 행번호 : {token.line_number} |"
              f"토큰이 몇번째 문장에 있는지 : {token.sent_position} |"
              f"토큰이 몇번째 어절인지 : {token.word_position}")
    
    print(result)
    print("-------------------------------------")

# 몇번째 어절이라는 건 ➔ 문장 내에서 몇번째 어절인가임. ! 


import pandas as pd

def tokentodf(tokens):
    """
    Token 객체 리스트를 받아 가독성이 좋은 Pandas DataFrame으로 변환합니다.
    """
    # 1. 각 토큰의 속성들을 딕셔너리 형태로 리스트에 담기
    data = []
    for token in tokens:
        token_info = {
            "토큰문자열": token.form,
            "토큰 원형": token.lemma,
            "토큰 품사": token.tag,
            "문서 내 시작 위치": token.start,
            "토큰 글자수": token.len,
            "행번호": token.line_number,
            "문장 번호": token.sent_position,
            "문장 내 어절 위치": token.word_position
        }
        data.append(token_info)
    
    # 2. 데이터프레임 생성 및 반환
    df = pd.DataFrame(data)
    return df
```

보통은 form 보다는 lemma( 원형 )으로 뽑아서 좀 더 중복을 제거하기 위해 사용함.

```python
# 문서를 문장단위로 분리. slit_into_sents
doc = '폴 바버 브라이튼 CEO는 영국 매체 <토크스포츠>와 인터뷰에서 토트넘의 제안을두번이나 거절했다고 전했다. 또한 바버 CEO는 판 헤케의 계약기간이1년밖에남지 않았으나 쉽게 타 팀으로 내줄생각이 없다는점도 분명히 했다. 바버 CEO는 "최고 수준 선수들에게 관심이 쏠리는 건 당연하고, 판 헤케 역시 마찬가지"라고 운을 뗀 후, "판 헤케를 향한 관심은 오래전부터 이어져 왔고 여러 구단에서 접근하고 있다"라고 시장에서 판 헤케가 많은 인기를 끌고 있다는 점은인정했다.'
```

```python
sent_list = kiwi.split_into_sents(doc)

sent_list
```

```python
# 글자만 가져오기
for sent in sent_list:
    # print(sent.text)
    print(kiwi.space(sent.text)) # 띄어쓰기 오류 바로잡기 
    print("-------------------------------------")
```
