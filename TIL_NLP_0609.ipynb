radius:5px;opacity:1.0"> Subword 방식의 tokenizers  library</mark>** 토크나이저를 생성할 수 있는 알고리즘 제공.
- 여러 llm에서 사용된 pretrained 토크나이|저도 제공 ➔ transformers library
    - Tokenizer 생성 : 토큰화 알고리즘을 지정해 instance 생성 <br>➔ 사전이 없는 토크나이저를 생성 ➔ 학습을 시켜야함(trainer를 통해서)
    trainer에게 dataset과 데이터 넣어서 학습시킨다. 
    - Trainer 생성 :
        - 역할 : 토크나이저를 학습해서 어휘사전(instance)을 생성 
        - 학습 파라미터 :
        - 대표적인 special token ([ ]나 < > 로 묶어줌)
            - [UNK] : OOV 토큰 (어휘사전에 존재하지 않는)
            - [PAD]: 패딩을 표시하는 토큰 ➔ 문서의 토큰 개수를 맞추기 위해서 개수가 모자랄 때 채우는 토큰
            - [SOS]: 문장의 시작을 표시하는 토큰
            - [EOS] : 문서의 종료를 표시하는 토큰
            - **[cls]**: 문서의 시작과 문서의 의미를 표현하는 토큰(BERT 모델이 사용하는 특수 토큰 )
            - [MASK] : 문서의 일부 내용들을 가리는 용도의 특수 토큰 ➔ 학습시킬 때 


```python
# 시간 측정을 위한 time
import time

from tokenizers import Tokenizer # 문서를 토큰화 하는 객체 * 토큰화는 어휘사전을 바탕으로
from tokenizers.models import BPE # 토큰화 Subword 알고리즘. (어떻게 토큰화할까? : 에서 역할별로 나눠서 정의해둠. )
from tokenizers.pre_tokenizers import Whitespace # 서브워드 방식으로 토큰화 하기 전에 문서를 미리 토큰화할 때 사용하는 방식
from tokenizers.trainers import BpeTrainer #Tokenizers를 학습시켜주는 트레이너 객체
```

```python
''' Tokenizer 생성 '''

tokenizer = Tokenizer(
    BPE(unk_token ='[UNK]') # UNK ➔ unknown 토큰을 등록해서 생성. : 즉 어휘사전에 없는 토큰
)

# pre tokenizer 추가 
tokenizer.pre_tokenizer = Whitespace() # 일단 이걸 가지고 나누고 시작 
```

```python
''' 학습을 위한 trainer 생성 '''

trainer = BpeTrainer(
    vocab_size=20_000, #어휘사전의 최대 단어 수 
    min_frequency=10, # 사전에 추가할 때 최소 출연 횟수(빈도수) ➔ 최소한 10번은 등장해야한다. 
    special_tokens=["[UNK]","[PAD]"], #어휘 사전에 추가할 특수 목적 토큰들 설정 ➔ unk_token에 설정한 것은 반드시 추가해야함.   
    continuing_subword_prefix= "##" # 연결 토큰과 어절 시작 토큰을 분리해서 저장. ➔ 연결 토큰 앞에 붙일 접두어를 설정 
)
```

**<mark style="background-color: #ffdf3d;color:black;border-radius:5px;opacity:1.0">🤔학습 파라미터 정리 </mark>** 
1. 국민청원이라는 한정적인 데이터셋 ➔ 스포츠 데이터에 대한 도메인이 맞지 않음.<br> 범용적으로 쓰기 위해서는 다양한 주제에 대한 학습data가 필요함
2. 기본 토큰이 8000개가 넘음, 근데 우리는 10000개로 한정 <br>➔ 실제 청원데이터로 만들어진 건 2000여건 정도 뿐. max_token의 한정 <br>
➔ vocab 사이즈를 키워야하지 않을까?

```python
''' Trainer 를 이용해서 Tokenizer를 학습(어휘사전 생성) '''

s = time.time()

# 학습시킬 문서 파일들의 경로를 리스트로 묶어서 전달
tokenizer.train(["data\petitions\petitions_corpus.txt"], trainer=trainer)

e = time.time()

print(f"걸린 시간:{e-s}초")
```

```python
# 학습된 정보를 저장 ➔ 토크나이저의 어휘사전 + 토크나이저관련설정들(json 포맷으로 저장됨)
# 저장할 폴더 만들고, 경로 지정 
import os

os.makedirs("save_model/tokenizers", exist_ok = True)
save_path ="save_model/tokenizers/petitions_bpe.json"

tokenizer.save(save_path)
```

- **<mark style="background-color:#FFC300;color:white;border-radius:5px;opacity:1.0">BPE와 WordPiece의 근본적인 개념</mark>** 단어를 잘게 쪼개서 사전(vocab)을 만든다.**
- WordPiece (확률 기반)
    - 두 글자 각각의 확률에 비해 / 그 둘이 붙어 다닐 확률이 얼마나 높은가
    - 연결 토큰임을 명시 : `##단어`
    - `소년`과 `##소년`은 사전 상의 다른 단어 <br> 청소년이 ➔ 청 + `##소년`+`##이`


```python
from tokenizers.models import WordPiece # bpe
from tokenizers.trainers import WordPieceTrainer #BPE trainer 
from tokenizers import Tokenizer
from tokenizers.pre_tokenizers import Whitespace
import time
```

```python
# Tokenizer 생성 - 모델 (wordpiece) 넣어서 생서
w_tokenizer = Tokenizer(
    WordPiece(unk_token="<unk>")
)

# Pre tokenizer 설정 ➔ 공백기준으로 자르고 그 다음부터 알고리즘을 시작하겠다. 
w_tokenizer.pre_tokenizer = Whitespace()
```

```python
# Trainer 설정
w_trainer = WordPieceTrainer(
    vocab_size = 20_000,
    special_tokens=["<unk>","<pad>","<sos>","<eos>","<mask>"]
)
```

```python
# 학습 
s = time.time()

w_tokenizer.train(
    ["data/petitions/petitions_corpus.txt"], # 파일이 여러개일 경우를 감안해서 리스트로 제공
    trainer = w_trainer
)

print(f"걸린시간:{time.time()-s}초")
```

```python
# 인코딩 ids 목록을 다시 문자열로 반환 (디코딩)
result = w_tokenizer.decode(encoding.ids)
```

```python
# 문자열로 반환하는 과정에서 # 처리하기 
result2 = ""

for token_id in encoding.ids:
    token = w_tokenizer.id_to_token(token_id)
    if token.startswith("##"): # token이 즉 ## 으로 시작하면 ## 다음 인덱스부터 값 붙이기 
        result2 += token[2:]
    else:
        result2 += " "+token # 아니면 한 칸 띄우고 붙이기 

print(result2.strip())
```

**<mark style="background-color: #ffdf3d;color:black;border-radius:5px;opacity:1.0">🤔나름대로 해석하기 </mark>** 
1. token 이 `##` 으로 시작하면 ➔ ## 다음 인덱스부터(딱 단어 시작부터) ➔ `token[2:]` 값을 바로 붙여버림 (띄어쓰기 없이 즉 앞 단어에 붙임)
2. 만약 `##`으로 시작하지 않는다면 ➔ 그자체로 하나의 토큰(어절) ➔ 한 칸 띄우고 (띄어쓰기) 붙임

<div style="color:white;display:fill;border-radius:8px;background-color:#323232;font-size:150%; letter-spacing:1.0px"><p style="padding: 12px;color:white;"><b><b><span style='color:white'><span style='color:#F1A424'>1.6 | </span></span></b>Unigram 토크나이저 학습</b></p></div>

- Unigram() 의 경우 따로 unk_token 설정하지 않는다
- Trainer의 special Token 중에 **0번 토큰을 unk_Token으로 사용한다.** 

```python
from tokenizers import Tokenizer
from tokenizers.models import Unigram
from tokenizers.trainers import UnigramTrainer
from tokenizers.pre_tokenizers import Whitespace
import time 
```

```python
# 토크나이저 객체 생성 
u_tokenizer = Tokenizer(
    Unigram()
)

u_tokenizer.pre_tokenizer = Whitespace()
u_trainer = UnigramTrainer(
    vocab_size = 20_000,
    special_tokens=["<unk>","<pad>"],
    min_frequency = 10
)
```

```python
# 학습
s = time.time()

u_tokenizer.train(
    ["data/petitions/petitions_corpus.txt"],
    trainer=u_trainer
)

print(f"걸린시간:{time.time()-s}초")
```

```python
# 저장
save_path = "save_model/tokenizers/petitions_unigram.json"
u_tokenizer.save(save_path)
```
