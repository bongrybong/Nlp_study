- 문서를 구성하는 단어들이 몇 번 나왔는지를 표현하는 행렬(matrix)로 벡터화 ( 테이블 형태)
- DTM / TDM (행과 열에 무엇을 두는가의 차이)
    - 행 : 문서 번호(Doc)
    - 열 : 모든 문서에서 나온 단어(Term)
    - 값 : 출연 빈도 수  
- 동일한 정보를 담고 있으며 두 행렬에 담긴 값만 서로 바뀐 형태임.
- 단점 :
    - 100개 문서의 모든 단어들이 다 들어감 ➔ 0이 너무 많은 희소행렬이됨.
    - 딥러닝보다는 머신러닝에서 주로 사용됨 
- 구현방법:
    1. 어휘사전(vocab) 생성 : 전체 문서에 나온 단어 수집하고 단어마다 고유 index 부여
    2. 문장과 단어간의 dtm 만들기
- 사이킷런 활용 : `Scikit-learn CountVectorizer`
- 파라미터 :
    - `stop_word`
    - `max_df` : int(횟수), float(비율) ➔ 특정 횟수 이상 나오는 것은 무시하도록 설정
    - `min_df`: 최소한 이정도는 나와야한다. (무시할 횟수/ 무시할 비율)
    - max_features: 최대 token수 : 빈도수가 높은 순서대로 정렬 ➔ 지정한 max_features 개수만큼 사용

```python
# 문서리스트
## 리스트의 각 원소들이 각 문서(doc)에 해당함. ➔ 기사 1개, 논문 1개 등.

text = ["나는 야구를 좋아합니다. 나는 축구를 좋아하지 않습니다.", # - doc1
        "친구는 야구와 축구를 좋아합니다. 야구보다 축구를 더 좋아합니다.", #-doc2
        "나는 축구를 더 좋아합니다. 영국 축구 리그보다 이탈리아 축구 리그를 더 좋아합니다.", #- doc3
        "나는 농구를 좋아하지 않습니다. 배구도 좋아하지 않습니다.", #- doc4
        "나는 액션영화를 좋아합니다.", #- doc5
        "어제 비빕밥을 먹었습니다."]#- doc6

stop_words = list('은는이가을를도.,') # 대표적인 조사들과 마침표를 리스트로 부여

print("문서개수: ", len(text))
print(stop_words)
```

```python
from kiwipiepy import Kiwi

a =  Kiwi()

a.tokenize(text[0]) # 토큰별로 잘라서 token 객체에 담아서 제공 
# 토큰의 원형값을 뽑아주는 것 : lemma
[token.lemma for token in a.tokenize(text[0])] # 이렇게 모인 값들 중에 stopwords 에 존재하는 경우 빼자 
[token.lemma for token in a.tokenize(text[0]) if token.form not in stop_words]
```

```python
# 전처리 및 토큰화 ➔ 형태소 기반 토큰화 진행
from kiwipiepy import Kiwi

kiwi =  Kiwi()

def tokenizer(doc:str) -> list:
    '''문서를 받아서 형태소 기반으로 토큰화해서 토큰 목록을 리스트로 반환'''
    tokens = kiwi.tokenize(doc) # 토큰리스트를 제공 : list[token]
    
    # 불용어 제외 & token 객체에서 원형값(lemma)만 추출해서 리스트로 저장
    tokens = [token.lemma for token in tokens if token.form not in stop_words]
    return tokens

tokenizer(text[0])
```

```python
# 사이킷런의 CountVectorizer를 활용해서 text to vector 작업 실행
## 문서의 의미를 표현하는 숫자 벡터로 변환
import sklearn

# 모델 출력 방식을 'text'로 변경 (기본값은 'diagram')
sklearn.set_config(display="text")

from sklearn.feature_extraction.text import CountVectorizer #사이킷런은 크게 변환기(전처리용 거의 다른 값으로 바꿔주는 애들_)와 추론기 : 모델 클래스들  그 중에서 countvectorizer 는 변환기

cv = CountVectorizer(
    #문서(str)을 받아서 토큰화하는 함수(callerable 설정) 생략하게 될 경우 token pattertn의 파라미터 설정을 따르게 됨. (기본 공백 / 구두점)
    token_pattern=None,
    tokenizer=tokenizer 
)

# 학습(fit)
cv.fit(text) # 입력: list[str-문서] ➔ fit(); 단어(토큰)들 추출
```

```python
dtm = cv.transform(text)
dtm
```

```python
arr = dtm.toarray()
print(arr.shape)
arr
```

</span></span></b>TF-IDF(Term Frequency|inverse document frequency)</b></p></div>

- 개별 문서에 많이 나오는 단어가 높은 값을 가지도록 함. 동시에 여러 문서에서 자주 나오는 단어에는 페널티 
- 모든 문서에서 단어가 많이 등장('을','를',' 야구') <br> 하는 경우 언어 특성. 주제상 많이 등장하는 단어임.
- 구현방법:
    - TF(Term Frequency) : 해당 단어가 해당 문서에서 몇번 나오는지
    - DF(Documnet Frequency) : 해당 단어가 몇개의 문서에서 나오는지
    - IDF : DF의 역수 (전체문서수 / 해당 단어가 나오는 문서수)
    - TF IDF의 정의 : TF*log idf
- 사이킷런의 `TfidfVectorizer`


```python
from sklearn import set_config
set_config(display='text')
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    tokenizer= tokenizer   
)

tfidf.fit(text)
```

```python
# 변환하기
f_vector = tfidf.transform(text)
f_vector.toarray()
```

```python
tf_idf = pd.DataFrame(
    f_vector.toarray(),
    columns=tfidf.get_feature_names_out(),
    index = [f"{i+1}번 문서" for i in range(arr.shape[0])], # shape의 0번 ➔ 데이터 개수
)
```
