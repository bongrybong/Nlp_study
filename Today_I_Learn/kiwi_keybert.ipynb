```python
from keybert import KeyBERT
from kiwipiepy import Kiwi
from transformers import BertModel
```

```python
model = BertModel.from_pretrained('skt/kobert-base-v1')
```

```python
kw_model = KeyBERT(model)
```

```python
text = """서울 지진 대피소에 대한 데이터 분석을 위해서는 어떤 종류의 데이터가 필요할까요? 예를 들어, 서울시의 지진 대피소 위치, 수용 가능 인원, 대피소 내부 시설물, 대피소 이용 현황 등의 정보가 필요할 것입니다. 지진 대피소 위치 분석 예시: 지진 대피소 위치는 서울시 공공데이터 포털에서 제공하는 "서울시 지진대피소 안내" 데이터를 사용할 수 있습니다. 이 데이터셋에는 지진 대피소 명칭, 위치(주소), 좌표, 수용 인원, 관리 기관 등의 항목이 포함되어 있습니다. 이를 바탕으로 대피소 위치를 지도에 시각화하여 지진 발생 시 대피소가 필요한 지역을 파악할 수 있습니다. 대피소 이용 현황 분석 예시: 대피소 이용 현황은 서울시에서 제공하는 "서울시 재난정보 실시간 수집 및 제공 서비스" 데이터를 사용할 수 있습니다. 이 데이터셋에는 대피소 이용 현황(대피소 이용 가능 여부, 이용 중인 인원 수), 지진 발생 시 대피소 이용 현황 등의 정보가 포함되어 있습니다. 이를 바탕으로 대피소 이용 현황을 분석하여 인원이 많은 대피소를 파악하거나, 대피소 이용 가능 여부 등을 파악할 수 있습니다."""
```

```python
# 형태소분석기 가져오기
# 키워드 추출하는 게 목표라서 문장 자체를 명사로만 이루어지게 재구성

kiwi = Kiwi()
result_text = ''
for sentence in kiwi.analyze(text):
    nouns = []
    for token in sentence[0]:
        if token.tag.startswith('NN'):
            nouns.append(token.form)
    if nouns:
        result_text = ' '.join(nouns)
print(result_text)
```

```python
# 키워드 추출
## ver1 . 한 어절만
keywords1 = kw_model.extract_keywords(result_text,
                                     keyphrase_ngram_range=(1,1),
                                     stop_words= None,
                                     top_n=10
                                     )

## ver2. 두어절까지
keywords2 = kw_model.extract_keywords(result_text,
                                     keyphrase_ngram_range=(1,2),
                                     stop_words= None,
                                     top_n=10
                                     )
```

- `keyphrase_ngram_range=(1, 1)` : 추출할 단어의 길이  추출할 키워드의 최소 및 최대 단어 글자(어절)수 지정 ➔ 튜플형태 (min_n,max_n) <br> ➔ 어절. 띄어쓰기 기준 단어의 개수를 의미 
- (1,1) : 1개 어절로만 이루어진 단어를 추출 <br> '자연','모델','성능
- (1,2) : 1개 어절부터 2개 어절로 이어진 구문까지 추출 <br> '자연어', 'transformer 모델'

- `top_n = 20` : 출력할 결과 계수  가장 중요도가 높은 상위 N개만 가져오기
- 정리:
현재 코드는 "원본 문서(result_text)에서 불용어 제거 없이(stop_words=None), 띄어쓰기 없는 단 단어 형태의 키워드만(keyphrase_ngram_range=(1,1)), 가장 연관성이 높은 순으로 20개(top_n=20) 뽑아줘" 라는 의미가 됩니다.
